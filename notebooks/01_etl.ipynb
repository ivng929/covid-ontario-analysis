```python
import pandas as pd
import numpy as np
```

## 1. Extraction

```python
# load dataset
df = pd.read_csv("../data/covid_2020.csv")
df.head()
```

```python
df.info()
```

```python
# check null values
df.isnull().sum()
```

## 2. Transform

```python
# standardize column names
df.columns = df.columns.str.lower()
```

```python
# handle missing values
df["test_reported_date"] = df["test_reported_date"].fillna("UNKNOWN")
df["specimen_date"] = df["specimen_date"].fillna("UNKNOWN")
df["outcome1"] = df["outcome1"].fillna("NON-FATAL")
```

```python
df.head()
```

## 3. Load

```python
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
# read credentials
from urllib.parse import quote_plus
DB_USER = os.getenv("DB_USER")
DB_PASSWORD = quote_plus(os.getenv("DB_PASSWORD"))
DB_HOST = os.getenv("DB_HOST")
DB_PORT = os.getenv("DB_PORT")
DB_NAME = os.getenv("DB_NAME")
```

```python
# connect to postgres
from sqlalchemy import create_engine, text
engine = create_engine(f"postgresql+psycopg2://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}")
```

```python
# create schema
with open("../sql/01_create_schema.sql", "r") as file:
    sql_script = file.read()
with engine.begin() as conn:
    conn.execute(text(sql_script))
```

```python
# load data into postgres
df.to_sql("covid_cases", 
          engine, 
          if_exists="replace",
          index=False,
          chunksize=20000,
          method="multi")
```
